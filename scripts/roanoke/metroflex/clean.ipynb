```python
import pathlib

import pandas as pd

from transit import cleaning
```

```python
project_root = pathlib.Path('../../..')
project_root.resolve()
```

```python
path_output = project_root / 'web/va/roanoke/metroflex'
path_output.mkdir(exist_ok=True, parents=True)
path_output.resolve()
```

```python
src_df = pd.read_excel(
    project_root / 'data/metroflex/Copy of FEBRUARY 2025 METRO FLX TRIP REPORT.xls',
    sheet_name='Sheet1',
    header=2
)
src_df
```

```python
df = cleaning.drop_empty_rows(src_df)
df.drop(df.index[-1], inplace=True) # drop last row
df.rename(columns={'Trip Date': 'Pickup Address Details'}, inplace=True)

df = cleaning.downshift_alternating_values(df, column='Customer', target='Trip Date', increment=2)
df = cleaning.drop_empty_rows(df)

df.drop([
    'Customer',
    'Dropoff Address',
    'Unnamed: 8',
    'Unnamed: 9',
    'Unnamed: 10',
    'Unnamed: 11',
    'Unnamed: 12',
    'Unnamed: 14',
    'Unnamed: 15',
    'Unnamed: 16',
    'Unnamed: 17',
], axis=1, inplace=True)

df.rename(columns={
    'Customer Summary': 'No show',
    'Miles': 'Guests',
    'Unnamed: 13': 'Attendants',
    'Pickup Address': 'Dropoff Address Details',
    'Pickup Address.1': 'Pickup Address',
    'Dropoff Address.1': 'Dropoff Address',
}, inplace=True)

df = df[df['No show'] == 'No']

df.to_csv(project_root / 'data/metroflex/metroflex-2025-02-trip-report-cleaned.csv')
df
```
